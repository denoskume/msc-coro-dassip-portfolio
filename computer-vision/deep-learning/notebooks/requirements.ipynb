## 2. Required Data

MNIST is not stored as one file per image. The complete training and test sets are packed into four canonical IDX files, so a 70,000-image dataset still requires only four data paths.

````python
# python-mnist expects these four canonical IDX containers.
DATA_DIR = Path("../data")
REQUIRED_MNIST_FILES = {
    "train-images-idx3-ubyte",
    "train-labels-idx1-ubyte",
    "t10k-images-idx3-ubyte",
    "t10k-labels-idx1-ubyte",
}

available = {
    path.name
    for path in DATA_DIR.glob("*-ubyte")
    if path.is_file()
}

missing = sorted(REQUIRED_MNIST_FILES - available)

assert DATA_DIR.is_dir(), f"MNIST data directory not found: {DATA_DIR}"
assert not missing, f"Missing MNIST IDX files: {missing}"

print(f"IDX files discovered: {len(available)}")
print("Required MNIST containers: 4/4")
````
