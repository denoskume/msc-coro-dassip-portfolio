## Context

Spatial-domain filtering modifies an image through local neighborhood operators. The resulting behavior depends jointly on kernel definition, normalization, boundary conditions, numerical precision, image noise characteristics, and the scale of relevant structures.

The laboratory addresses both restoration-oriented filters and derivative/sharpening operators, with emphasis on controlled comparison between filtering objectives such as noise suppression, edge preservation, detail enhancement, and gradient estimation.

**Problem.** Design, implement, and evaluate spatial-domain filtering operations for smoothing, denoising, sharpening, and derivative estimation under explicit numerical and boundary conditions.

Frequency-domain filtering, segmentation, and learned convolutional models are outside scope.



## Learning objectives

By the end of this lab, you should be able to:

- Explain spatial convolution, kernel behavior, and boundary handling in image filtering.
- Apply smoothing and denoising filters and compare their effect on noise and detail.
- Perform sharpening and gradient-based edge extraction with standard spatial operators.
- Compare filters using quantitative measures together with visual evidence.
- Validate a reproducible spatial-filtering workflow and its generated outputs.
