````python
import sys
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
import numpy as np

print("Python:", sys.version.split()[0])
print("Feature-tracking libraries imported successfully.")
````

## 2. Required Data

Tracking videos are discovered recursively from `../data/`. The implementation processes one sequence at a time by selecting an index from the discovered video list, so adding videos requires no new path definitions.

````python
# Mirror the dynamic video discovery used by main.ipynb.
DATA_DIR = Path("../data")
VIDEO_EXTENSIONS = {".mp4", ".avi", ".mov", ".mkv"}

video_paths = sorted(
    path for path in DATA_DIR.rglob("*")
    if path.is_file() and path.suffix.lower() in VIDEO_EXTENSIONS
)

assert video_paths, f"No supported video found in: {DATA_DIR}"

VIDEO_INDEX = 0
video_path = video_paths[VIDEO_INDEX]

capture = cv2.VideoCapture(str(video_path))
opened = capture.isOpened()
read_ok, first_frame = capture.read()
capture.release()

assert opened, f"Could not open video: {video_path}"
assert read_ok and first_frame is not None, "Could not decode the first frame."

print(f"Videos discovered: {len(video_paths)}")
print(f"Selected video   : {video_path.name}")
````
