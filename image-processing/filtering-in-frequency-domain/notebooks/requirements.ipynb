## 2. Required Data

The dataset is discovered dynamically by subfolder. `Fourier/`, `PhaseMag/`, and `Frequency/` may grow without adding filename lines; the notebook resolves only the semantic examples needed by specific experiments.

````python
# Discover supported images by dataset group, exactly as main.ipynb does.
IMAGE_EXTENSIONS = {".png", ".jpg", ".jpeg", ".tif", ".tiff", ".bmp"}

def find_lab_root(start: Path) -> Path:
    for candidate in [start.resolve(), *start.resolve().parents]:
        if (
            (candidate / "data").is_dir()
            and (candidate / "notebooks" / "main.ipynb").is_file()
        ):
            return candidate
    raise FileNotFoundError("Could not locate the frequency-domain lab root.")


LAB_ROOT = find_lab_root(Path.cwd())
DATA_DIR = LAB_ROOT / "data"

DATASETS = {
    folder.name: sorted(
        path for path in folder.rglob("*")
        if path.is_file() and path.suffix.lower() in IMAGE_EXTENSIONS
    )
    for folder in DATA_DIR.iterdir()
    if folder.is_dir()
}

for group in ("Fourier", "PhaseMag", "Frequency"):
    assert DATASETS.get(group), f"No supported images found in data/{group}/"

def has_role(group: str, keyword: str) -> bool:
    keyword = keyword.lower()
    return any(keyword in path.stem.lower() for path in DATASETS[group])

for group, keyword in [
    ("Fourier", "house"),
    ("PhaseMag", "cat"),
    ("PhaseMag", "wolf"),
    ("Frequency", "astronaut"),
    ("Frequency", "moire"),
    ("Frequency", "spotshade"),
]:
    assert has_role(group, keyword), f"Missing role '{keyword}' in data/{group}/"

for group in ("Fourier", "PhaseMag", "Frequency"):
    print(f"{group:10s}: {len(DATASETS[group])} image(s)")
````
