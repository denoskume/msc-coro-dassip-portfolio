## Context

Centrale Nantes — MATSIP Part I, **MATLAB only**. This lab studies nonlinear descent methods, denoising and NMR nonnegative inverse problems. The MATLAB implementation was run on real NMR data; the twelve plots have been reviewed. This module preserves the original professor-required programming language and provides the same four-notebook portfolio structure.

**Outputs:** MATLAB scripts, helper functions, authentic course NMR data, 12 saved MATLAB figures, numerical diagnostics and scientific interpretation.
