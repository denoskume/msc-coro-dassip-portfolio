## 2. Required Data

Segmentation images are discovered recursively from `data/`. The notebook selects the current experiment roles (`hand`, `tower`, `peppers`) from that discovered set, so adding hundreds or thousands of images does not add path lines.

````python
IMAGE_EXTENSIONS = {".png", ".jpg", ".jpeg", ".tif", ".tiff", ".bmp"}

def find_lab_root(start: Path) -> Path:
    for candidate in [start.resolve(), *start.resolve().parents]:
        if (
            (candidate / "data").is_dir()
            and (candidate / "notebooks" / "main.ipynb").is_file()
        ):
            return candidate
    raise FileNotFoundError("Could not locate the segmentation lab root.")


LAB_ROOT = find_lab_root(Path.cwd())
DATA_DIR = LAB_ROOT / "data"

DATA_FILES = sorted(
    path for path in DATA_DIR.rglob("*")
    if path.is_file() and path.suffix.lower() in IMAGE_EXTENSIONS
)

assert DATA_FILES, f"No supported images found in: {DATA_DIR}"

ROLE_NAMES = ("hand", "tower", "peppers")

missing_roles = [
    role for role in ROLE_NAMES
    if not any(
        path.stem.lower() == role
        or role in path.stem.lower()
        for path in DATA_FILES
    )
]

assert not missing_roles, f"Missing segmentation roles: {missing_roles}"

print(f"Images discovered: {len(DATA_FILES)}")
print(f"Experiment roles : {len(ROLE_NAMES)}")
````
