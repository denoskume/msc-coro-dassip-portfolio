## Context

Digital image processing operates on sampled, quantized arrays whose numerical representation directly affects every downstream operation. Image dimensions, channel order, data type, dynamic range, coordinate convention, and encoding format are therefore part of the computational model rather than incidental implementation details.

This laboratory establishes a controlled numerical framework for inspecting image data, manipulating pixels and regions, characterizing intensity distributions, simulating common noise processes, and comparing images using reproducible quantitative measures.

**Problem.** Characterize and validate the numerical representation of binary, grayscale, and RGB images and establish a reproducible set of core image-processing operations.

Geometric transformation, spatial/frequency filtering, segmentation, and projective vision are handled in dedicated modules and are outside this laboratory's scope.



## Learning objectives

By the end of this lab, you should be able to:

- Explain how digital images are represented through sampling, quantization, pixels, channels, and data types.
- Inspect and manipulate image arrays safely using coordinates, regions of interest, and channel operations.
- Analyze intensity distributions using statistics, histograms, and dynamic-range transformations.
- Simulate common noise models and compare images with standard numerical error metrics.
- Apply a consistent inspection workflow to validate image properties and processing results.

## Input data

| Item | Fixed value / convention |
| --- | --- |
| Data directory | `../data/` |
| Reference images | `Elizabeth_Tower_London.jpg`, `ballons.jpg`, `einstein.png`, `grass.jpg`, `peppers.png` |
| Primary libraries | NumPy, Pillow, Matplotlib |
| Color convention | RGB unless explicitly stated otherwise |
| Grayscale convention | one intensity channel |
| Standard 8-bit range | $0$–$255$ |
| Floating-point normalized range | $[0,1]$ when explicitly used |
| Histogram bins for 8-bit grayscale | 256 |
| Generated figures | `../outputs/figures/` |
| Randomness | fixed notebook seed for synthetic noise |

All numerical operations must preserve dtype/range semantics explicitly. Any conversion between integer and floating-point representations must be deliberate and documented.
