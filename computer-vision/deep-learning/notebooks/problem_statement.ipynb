## Context

Handwritten-digit recognition provides a controlled setting for evaluating how model capacity influences supervised classification performance under a fixed training protocol. MNIST contains 28×28 grayscale images from ten digit classes and is used here to compare multilayer perceptrons under identical data, optimization, and evaluation conditions.

The experiment is designed as a capacity study rather than an unrestricted model search. Dataset handling, preprocessing, optimizer, loss function, batch size, learning rate, weight decay, and training duration remain fixed so that hidden-layer width is the principal experimental variable.

**Problem.** Train and compare three one-hidden-layer PyTorch MLP classifiers with hidden widths of 128, 256, and 512 neurons on the same MNIST training/test split.

The comparison is limited to the stated MLP family; CNNs, augmentation, hyperparameter search, and separate calibration procedures are outside scope.



## Learning objectives

By the end of this lab, you should be able to:

- Prepare and validate the MNIST dataset for reproducible neural-network experiments.
- Build and train a one-hidden-layer MLP classifier with a consistent optimization pipeline.
- Compare model capacity by evaluating multiple hidden-layer sizes under the same protocol.
- Interpret predictions, confidence scores, and confidence-threshold behavior.
- Validate numerical results and generated evaluation figures before drawing conclusions.

## Input data

| Item | Fixed value |
| --- | --- |
| Dataset | MNIST local IDX files |
| Training set | 60,000 images |
| Test set | 10,000 images |
| Image size | $28\times28$ grayscale |
| Flattened input dimension | $784$ |
| Number of classes | $10$ |
| Hidden widths | $128$, $256$, $512$ |
| Hidden block | Linear → BatchNorm1d → ReLU |
| Output | 10 logits |
| Loss | Cross-Entropy |
| Optimizer | Adam |
| Epochs | 10 |
| Batch size | 512 |
| Learning rate | $10^{-3}$ |
| Weight decay | $10^{-3}$ |
| Confidence thresholds | 100 values uniformly spanning $[0,0.99]$ |
| Required outputs | six diagnostic figures under `outputs/figures/` |

The dataset split, preprocessing, optimizer, loss, and training duration are fixed for all three architectures.
