This notebook checks only the dependencies and input data actually required by `main.ipynb` for the camera-calibration experiment.

## 1. Software Requirements

- Python 3.12
- NumPy
- Matplotlib
- OpenCV
- IPython/Jupyter kernel

Exact package versions are recorded in `../requirements.txt`.

````python
import sys
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
import numpy as np

print("Python:", sys.version.split()[0])
print("Camera-calibration libraries imported successfully.")
````

## 2. Required Data

The calibration dataset is discovered dynamically from `../data/calibration_images/`. Any supported calibration image added to this folder is included automatically; no filename list is maintained.

````python
# Mirror the dynamic discovery used by main.ipynb.
DATA_DIR = Path("../data/calibration_images")
IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff"}

image_paths = sorted(
    path for path in DATA_DIR.iterdir()
    if path.is_file() and path.suffix.lower() in IMAGE_EXTENSIONS
)

assert DATA_DIR.is_dir(), f"Calibration directory not found: {DATA_DIR}"
assert image_paths, f"No supported calibration images found in: {DATA_DIR}"

readable = [path for path in image_paths if cv2.imread(str(path)) is not None]
assert readable, "No readable calibration image was found."

print(f"Calibration files found: {len(image_paths)}")
print(f"Readable views        : {len(readable)}")
````
