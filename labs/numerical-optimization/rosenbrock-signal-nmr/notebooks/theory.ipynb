## 1. Rosenbrock Function, Gradient, Hessian and Optimum

Rosenbrock with b=2 has global minimum (1,1) and objective zero. The Hessian at the minimizer is positive definite: [[18,-8],[-8,4]].

**À retenir :** Rosenbrock f*=0 at (1,1); Hessian is positive definite there.


## 2. Rosenbrock Surface and Level Sets

Visualize log(1+f) as 3D surface and contour levels, with the unique minimum marked.

**À retenir :** A narrow, curved valley challenges naive gradient updates.


## 3. Fixed-Step Gradient Descent

Four fixed steps show the difference between oscillatory large steps, reasonable steps, and very slow small steps.

**À retenir :** Too-large steps oscillate; too-small steps slow progress.


## 4. Armijo Backtracking

Armijo backtracking chooses a step ensuring sufficient descent; the numerical example converged to (1,1).

**À retenir :** Armijo gives sufficient decrease, not necessarily exact line search.


## 5. Six-Method Optimization Comparison

Compare Gradient, Fletcher–Reeves, Newton, BFGS, Gauss–Newton and Levenberg–Marquardt from two starts, using objective, gradient norm, iterations, time and paths.

**À retenir :** On this example Gauss–Newton took 4–5 iterations, Gradient Descent 495–530.


## 6. Simulate the Signal and Add Noise

Generate 1,000 synthetic signal samples, then Gaussian noise with target SNR=20 dB.

**À retenir :** 20 dB SNR = signal power 100 times noise power.


## 7. Tikhonov Signal Enhancement

Minimize fidelity plus quadratic smoothness using closed-form linear algebra and Newton iterations; compare the results.

**À retenir :** Quadratic inverse smoothing can be solved with a linear system.


## 8. Select the Regularization Weight

Sweep lambda and compare estimation error, fidelity, smoothness; simulation best lambda=0.5554.

**À retenir :** Bias-variance/fidelity-smoothness trade-off selects the weight.


## 9. Synthetic NMR Exterior-Penalty Inversion

Minimize NMR residual norm with nonnegativity via quadratic exterior penalty; distinguish constraint feasibility from recovery of the known distribution.

**À retenir :** Feasibility passed but synthetic NMR distribution error was 300%.


## 10. Experimental NMR Exterior-Penalty Inversion

Repeat penalty method on authentic measured NMR data; do not infer physical ground truth from fit alone.

**À retenir :** Measured fit error was 2.360%, but no true distribution is available.
