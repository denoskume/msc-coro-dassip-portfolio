## 2. Required Data

All images are discovered recursively from `data/`, including nested bilateral-filter folders. The code then resolves only the semantic roles required by the experiments; adding unrelated or additional images requires no path changes.

````python
IMAGE_EXTENSIONS = {".png", ".jpg", ".jpeg", ".tif", ".tiff", ".bmp"}

def find_lab_root(start: Path) -> Path:
    for candidate in [start.resolve(), *start.resolve().parents]:
        if (candidate / "data").is_dir() and (candidate / "notebooks").is_dir():
            return candidate
    raise FileNotFoundError("Could not locate the spatial-filtering lab root.")


LAB_DIR = find_lab_root(Path.cwd())
DATA_DIR = LAB_DIR / "data"

DATA_FILES = sorted(
    path for path in DATA_DIR.rglob("*")
    if path.is_file() and path.suffix.lower() in IMAGE_EXTENSIONS
)

assert DATA_FILES, f"No supported images found in: {DATA_DIR}"

def has_role(*tokens: str) -> bool:
    tokens = tuple(token.lower() for token in tokens)
    return any(
        all(
            token in path.relative_to(DATA_DIR).as_posix().lower()
            for token in tokens
        )
        for path in DATA_FILES
    )

ROLE_PATTERNS = {
    "moon": ("moon-blurred",),
    "ascent": ("ascentb",),
    "einstein_ref": ("einstein/ref",),
    "einstein_gaussian": ("einstein_gaussian",),
    "einstein_salt": ("saltandpepper",),
    "einstein_speckle": ("speckle",),
    "taj_ref": ("tajmahal/ref",),
    "zebra_ref": ("zebra/ref",),
    "monarch_ref": ("monarch/ref",),
}

missing_roles = [
    role for role, tokens in ROLE_PATTERNS.items()
    if not has_role(*tokens)
]

assert not missing_roles, f"Missing experiment roles: {missing_roles}"

print(f"Images discovered : {len(DATA_FILES)}")
print(f"Experiment roles  : {len(ROLE_PATTERNS)}")
````
